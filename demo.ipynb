# QuantItera — AI Agent That Validates Its Own Predictions

**Discover. Reason. Validate. Evolve.**

Run the official `QuantItera` closed-loop research and portfolio performance tutorials in one click — **no API keys required, zero permanent workspace pollution** (uses an isolated temporary directory that cleans up automatically).

- GitHub Repository: [https://github.com/wysking/quantitera](https://github.com/wysking/quantitera)

```python
import json
from pathlib import Path
import subprocess
import sys
import tempfile

REPO_URL = "https://github.com/wysking/quantitera.git"

with tempfile.TemporaryDirectory(prefix="quantitera_colab_") as tmp:
    tmp_path = Path(tmp)
    # If already running inside a local checkout of quantitera, reuse it; otherwise shallow-clone into temp dir
    cwd = Path.cwd()
    if (cwd / "research_engine").is_dir() and (cwd / "examples/research_demo.py").is_file():
        repo_dir = cwd
    else:
        repo_dir = tmp_path / "quantitera"
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, str(repo_dir)], check=True)

    sys.path.insert(0, str(repo_dir))
    from examples.research_demo import run_demo as run_research_demo
    from examples.performance_demo import run_demo as run_performance_demo

    research_ws = tmp_path / "tutorial-workspace"
    perf_ws = tmp_path / "performance-workspace"

    summary = run_research_demo(research_ws)
    perf_summary = run_performance_demo(perf_ws)

    research_report = (research_ws / "exports/tutorial/research_report.md").read_text(encoding="utf-8")
    research_perf_md = (research_ws / "exports/research_performance.md").read_text(encoding="utf-8")
    live_perf_md = (perf_ws / "exports/live_portfolio_performance.md").read_text(encoding="utf-8")

print("=== 1. CLI Summary (JSON) ===")
print(json.dumps({
    "example_only": True,
    "report": "exports/tutorial/research_report.md",
    "counts": summary["research_counts"],
    "repeat_review_events_written": summary["repeat_review_events_written"],
}, ensure_ascii=False, indent=2))
print("\n" + "=" * 72 + "\n")
print(research_report)
print("=" * 72 + "\n")
print(live_perf_md)
```
